## Dataset
The **Iris Dataset** contains 150 samples of iris flowers from 3 species:
- *Iris setosa*, *Iris versicolor*, *Iris virginica*

Each sample has 4 measurements: **sepal length**, **sepal width**, **petal length**, **petal width** (in cm).

## Step 1: Import Libraries

```python
# Core data libraries
import pandas as pd
import numpy as np

# Visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
sns.set_theme(style='whitegrid', palette='Set2')
plt.rcParams['figure.figsize'] = (10, 6)
pd.set_option('display.precision', 2)

import os
os.makedirs('plots', exist_ok=True)

print('All libraries loaded successfully!')
```

## Step 2: Load the Dataset

```python
# Load Iris dataset via seaborn (no download needed)
df = sns.load_dataset('iris')

print(f'Dataset Shape: {df.shape}')
print(f'Columns: {list(df.columns)}')
print(f'Species: {df["species"].unique().tolist()}')
```

## Step 3: Initial Inspection

```python
# First 5 rows
print('=== First 5 Rows ===')
df.head()
```

```python
# Data types and null counts
print('=== Dataset Info ===')
df.info()
```

```python
# Statistical summary
print('=== Descriptive Statistics ===')
df.describe()
```

```python
# Class balance check
print('=== Species Distribution ===')
print(df['species'].value_counts())

# Missing values check
print('\n=== Missing Values ===')
print(df.isnull().sum())
```

> **Insight:** The dataset is perfectly balanced — 50 samples per species — and contains **no missing values**. All 4 features are continuous numerical measurements.

## Step 4: Visualizations
### 4.1 Pairplot — All Feature Relationships at Once

```python
# Pairplot shows every feature combination, colored by species
g = sns.pairplot(df, hue='species', diag_kind='kde', plot_kws={'alpha': 0.7})
g.fig.suptitle('Pairplot of Iris Features by Species', y=1.02, fontsize=14, fontweight='bold')
plt.savefig('plots/pairplot.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: plots/pairplot.png')
```

### 4.2 Scatter Plot — Petal Length vs Petal Width

```python
# Petal features are the most discriminative — visualize them directly
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df, x='petal_length', y='petal_width',
    hue='species', s=100, edgecolor='black', linewidth=0.5
)
plt.title('Petal Length vs Petal Width by Species', fontsize=14, fontweight='bold')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.legend(title='Species')
plt.tight_layout()
plt.savefig('plots/scatter_petal.png', dpi=150)
plt.show()
```

> **Insight:** *Setosa* is completely separated from the other two species using just petal measurements. *Versicolor* and *Virginica* show slight overlap.

### 4.3 Histograms — Value Distributions per Feature

```python
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Feature Distributions by Species', fontsize=15, fontweight='bold')

for ax, feature in zip(axes.flatten(), features):
    for species in df['species'].unique():
        subset = df[df['species'] == species]
        ax.hist(subset[feature], bins=15, alpha=0.6, label=species, edgecolor='white')
    ax.set_title(feature.replace('_', ' ').title())
    ax.set_xlabel('Value (cm)')
    ax.set_ylabel('Frequency')
    ax.legend(title='Species', fontsize=8)

plt.tight_layout()
plt.savefig('plots/histograms.png', dpi=150)
plt.show()
```

> **Insight:** Petal length and petal width show clearly separated distributions per species. Sepal width distributions heavily overlap — it is the least useful standalone feature.

### 4.4 Box Plots — Spread and Outlier Detection

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Box Plots — Feature Spread and Outliers by Species', fontsize=15, fontweight='bold')

for ax, feature in zip(axes.flatten(), features):
    sns.boxplot(data=df, x='species', y=feature, ax=ax, palette='Set2')
    ax.set_title(feature.replace('_', ' ').title())
    ax.set_xlabel('Species')
    ax.set_ylabel('Value (cm)')

plt.tight_layout()
plt.savefig('plots/boxplots.png', dpi=150)
plt.show()
```

> **Insight:** A few mild outliers appear in sepal width for *Setosa*, but the dataset is overall very clean. Box plots confirm petal features have the smallest within-species spread, making them reliable predictors.

### 4.5 Correlation Heatmap

```python
plt.figure(figsize=(8, 6))
corr = df.drop('species', axis=1).corr()
sns.heatmap(
    corr, annot=True, fmt='.2f', cmap='coolwarm',
    square=True, linewidths=0.5, annot_kws={'size': 12}
)
plt.title('Feature Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('plots/heatmap.png', dpi=150)
plt.show()
```

> **Insight:** Petal length and petal width are very strongly correlated (r = 0.96). Sepal width shows weak or negative correlations with other features.
